```python
#linearRegression
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([1,2,3,4,5,6,7,8])
y = np.array([2,4,6,8,10,13,14,16])

# Parameters
lr = 0.01
w = 0
b = 0
n = len(x)
epochs = 1000

loss = []

# Training loop
for i in range(epochs):

    # Prediction
    y_pred = w * x + b

    # Error
    error = y_pred-y

    # Cost function (MSE)
    J = (1/(2*n)) * np.sum(error ** 2)
    loss.append(J)

    # Gradients
    dw = (1/n) * np.sum(x * error)
    db = (1/n) * np.sum(error)

    # Update weights
    w = w - lr * dw
    b = b - lr * db

# Final parameters
print("Weight (w):", w)
print("Bias (b):", b)

# Prediction example
X_test = 2
Y_predicted = w * X_test + b
print("Prediction for x=2:", Y_predicted)

# Plot loss curve
plt.plot(loss)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training Loss Curve")
plt.show()

# Plot regression line
plt.scatter(x, y)
plt.plot(x, w*x + b)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Fit")
plt.show()
```
